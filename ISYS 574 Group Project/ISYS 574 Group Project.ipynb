## Data Preparation

```python
# import libraries
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
```

```python
#Load the CSV
df = pd.read_csv("cybersecurity_intrusion.csv")
df.head()
```

### (Step 1) Standardize column names

```python
# Find column names
df.columns
```

```python
# Standardize all column names
# Remove leading/trailing spaces
# Convert to lowercase
# Replace spaces with underscores

df.columns = (df.columns.str.strip().str.lower().str.replace(" ", "_"))

# Display updated column names
print(df.columns.tolist())
```

```python
df
```

### (Step 2) Standardize Data

```python
for col in df.select_dtypes(include=['string', 'object']).columns:
    print(f"\n ({col}) ")
    print(df[col].unique())
```

```python
cat_cols = df.select_dtypes(include=['string', 'object']).columns

for col in cat_cols:
    df[col] = df[col].str.strip().str.upper()
    print(df[col].unique())
```

```python
df
```

### (Step 3) Handle missing values

```python
# Check missing values in each column
print(df.isnull().sum())
```

```python
# Fill Missing with 0 so that they dont affect the model.
df["protocol_type"] = df["protocol_type"].fillna(0)
df["encryption_used"] = df["encryption_used"].fillna(0)

# Verify missing values are gone
print(df.isnull().sum())
```

```python
df
```

Ways to handle missing categorical values:
1. Replace missing values with the most common category BUT this distorts category distribution
2. Replace missing values with "Unknown", this preserves the "missingness" of the data by introducing an additional category. BUT the dataset is NOT real. Therefore, I should not make any assumptions about the importance of some features compared to 
others. This could be helpfull in the real world because if we are not able to detect the protocal type or encription used it may be a good predictor of threat severity or detected attack. You would have to calculate r^2 and multicolnarity if you want to proceed with this option.
3. Remove observations with missing values BUT this makes us lose potentially valuable data.
4. Replace missing values by sampling observed categories, this may preserves category proportions BUT it introduces randomness and uncertainty.
5. Use other features to predict missing categories, this can capture relationships between features BUT it's more complex and may introduce errors.

### (Step 5) Check for Imposible observations

```python
numerical_columns = ["network_packet_size","login_attempts","session_duration","ip_reputation_score","risk_level","failed_logins","threat_severity_score"]
```

```python
# Check for negative values
for col in numerical_columns:
    print(col, "Negative values:", (df[col] < 0).sum())

# Check IP reputation score range (0 to 1)
print("Invalid IP Reputation Scores:",
      ((df["ip_reputation_score"] < 0) |
       (df["ip_reputation_score"] > 1)).sum())

# Check threat severity score range (0 to 100)
print("Invalid Threat Severity Scores:",
      ((df["threat_severity_score"] < 0) |
       (df["threat_severity_score"] > 100)).sum())

# Check risk level range (1 to 3)
print("Invalid Risk Levels:",
      (~df["risk_level"].isin([1, 2, 3])).sum())
```

### (Step 6) Encode Categorical Variables

```python
# One-hot encode categorical features
categorical_columns = ["protocol_type","encryption_used","browser_type"]

df = pd.get_dummies(df,columns=categorical_columns,dtype=int)
```

```python
# Encode binary variables
binary_columns = ["unusual_time_access","attack_detected"]

for col in binary_columns:
    df[col] = (df[col].astype(str).str.strip().str.lower().map({"yes": 1, "no": 0}))
```

```python
# Display updated dataset
print(df.head())

# Display column names
print(df.columns.tolist())

# Check data types
print(df.dtypes)

# Check for missing values after encoding
print(df.isnull().sum())
```

### Step 7 Split the Data

```python
# threat dataset
threat_df = df.drop(columns=["attack_detected"]).copy()

# attack dataset
attack_df = df.drop(columns=["threat_severity_score"]).copy()
```

```python
# Split into 80% training/validation and 20% testing
threat_df_full_train, threat_df_test = train_test_split(threat_df, test_size=0.2, random_state=5)

# Split into 60% training and 20% validation
threat_df_train, threat_df_val = train_test_split(threat_df_full_train, test_size=0.25, random_state=5)

# Check dataset sizes
print(len(threat_df_train), len(threat_df_val), len(threat_df_test))
```

```python
# Reset indices
threat_df_train = threat_df_train.reset_index(drop=True)
threat_df_val = threat_df_val.reset_index(drop=True)
threat_df_test = threat_df_test.reset_index(drop=True)

# Extract target variable
y_train_threat = threat_df_train.threat_severity_score.values
y_val_threat = threat_df_val.threat_severity_score.values
y_test_threat = threat_df_test.threat_severity_score.values

# Remove target from predictor datasets
del threat_df_train["threat_severity_score"]
del threat_df_val["threat_severity_score"]
del threat_df_test["threat_severity_score"]
```

```python
# Split into 80% training/validation and 20% testing
attack_df_full_train, attack_df_test = train_test_split(
    attack_df,
    test_size=0.2,
    random_state=5,
    stratify=attack_df["attack_detected"]
)

# Split into 60% training and 20% validation
attack_df_train, attack_df_val = train_test_split(
    attack_df_full_train,
    test_size=0.25,
    random_state=5,
    stratify=attack_df_full_train["attack_detected"]
)

# Check dataset sizes
print(len(attack_df_train), len(attack_df_val), len(attack_df_test))
```

```python
# Reset indices
attack_df_train = attack_df_train.reset_index(drop=True)
attack_df_val = attack_df_val.reset_index(drop=True)
attack_df_test = attack_df_test.reset_index(drop=True)

# Extract target variable
y_train_attack = attack_df_train.attack_detected.values
y_val_attack = attack_df_val.attack_detected.values
y_test_attack = attack_df_test.attack_detected.values

# Remove target from predictor datasets
del attack_df_train["attack_detected"]
del attack_df_val["attack_detected"]
del attack_df_test["attack_detected"]
```
